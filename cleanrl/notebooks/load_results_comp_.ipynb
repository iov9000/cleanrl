```python
def process_results(envs, results, param_filter_key_list, param_filter_value_list, n_demos=[1,4,10], sss=[1,20,100], seeds=[515, 234, 894, 322, 72], print_files=False, len_thr=None):
    # filter into airl, gail_lip
    d, d_mustd = {}, {}
    
    #envs = ['AdroitHandDoor-v1', 'AdroitHandPen-v1', 'FrankaKitchen-v1']

    for env_name in envs:
        d[env_name] = {}
        for seed in seeds:
            d[env_name][seed] = {}
            for n_demo in n_demos:
                d[env_name][seed][n_demo]  = {}
                for ss in sss:
                    d[env_name][seed][n_demo][ss]  = {}
                    
    
    for env_name in envs:
        d_mustd[env_name] = {}
        for n_demo in n_demos:
            d_mustd[env_name][n_demo]  = {}
            for ss in sss:
                d_mustd[env_name][n_demo][ss]  = {}
                
    # load results and filter them according to keyss
    for k, v in results.items():
        hparams = v[2]
        env_name = hparams['env_id']

        seed = int(hparams['seed'])
        n_d = int(hparams['n_demos'])
        if 'pwil' in str(k):
            ss = int(hparams['subsampling'])
        else:
            ss = int(hparams['subsample'])
        
        if not v[2]:
            print("empty dict", k)

        if print_files:
            print(k)
            
        filter_list = []
        for k_hp, v_hp in zip(param_filter_key_list, param_filter_value_list):

            if k_hp in hparams.keys():
                filter_list.append(hparams[k_hp] == v_hp)
        # if 'td3_swil' in str(k):
        #     print(k)
        #     print(hparams, param_filter_value_list)
        #     print(filter_list)
        if all(filter_list):
            #print("filters satisfied:", env_name, seed, n_d, ss)
            d[env_name][seed][n_d][ss] = v
            
    # fill mu,std dict by averagin over seeds
    print("Computing mu/std")
    for env in envs:
        for ss in sss:
            for n_d in n_demos:
                val_list = []
                min_len = 1000
                
                if len_thr is None:
                    if 'Humanoid' in env:
                        len_thresh = 80
                    else:
                        len_thresh = 90
                else:
                    len_thresh = len_thr

                print("loading hparams:", env, seed, n_d, ss)
                hparams = d[env][seed][n_d][ss][2]
                for seed in seeds:
                    if len(d[env][seed][n_d][ss][1]['Value']) >= len_thresh:
                        val_list.append(d[env][seed][n_d][ss][1]['Value'][:len_thresh])
                    else:
                        print(f"Incomplete length: {hparams['exp_name']}", env, seed, n_d, ss, len(d[env][seed][n_d][ss][1]['Value']))

                if len(val_list) >= 2:
                    mu = np.mean(np.stack(val_list),0)
                    std = np.std(np.stack(val_list),0)
                else:
                    mu, std = np.zeros(len_thresh), np.zeros(len_thresh)

                d_mustd[env][n_d][ss] = (mu, std)
    
    return d, d_mustd
                
```
